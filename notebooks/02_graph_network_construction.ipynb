# Phase 3: Graph Construction
### Mapping the 'Hydra' Network Connectivity

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# 1. Load the data
df = pd.read_csv('../data/raw/transactions.csv')

# 2. Build the Graph
# We treat 'user_id' as nodes. For this simulation, we'll assume 
# fraud bots are transacting within a specific subnet.
G = nx.from_pandas_edgelist(df.head(1000), source='user_id', target='transaction_id')

# 3. Calculate Degree Centrality
# This identifies 'High-Velocity' nodes that are too active to be human.
degrees = dict(G.degree())
top_nodes = sorted(degrees.items(), key=lambda x: x[1], reverse=True)[:5]

print('--- High-Velocity Node Analysis ---')
for node, count in top_nodes:
    print(f'Account {node} has {count} rapid-fire connections.')

# 4. Visualizing a slice of the network
plt.figure(figsize=(10,10))
nx.draw(G, node_size=20, node_color='skyblue', alpha=0.6)
plt.title('Transaction Graph: Identifying Non-Human Clusters')
plt.show()
```
